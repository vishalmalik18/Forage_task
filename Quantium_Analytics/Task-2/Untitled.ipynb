```python
from scipy.stats import pearsonr
import numpy as np

def compare_stores_metric(trial_df, control_df, metric='total_sales'):
    # Merge on month to align months
    merged = trial_df.merge(control_df, on='month', suffixes=('_trial', '_control'))

    if len(merged) < 2:
        print("Not enough overlapping months between trial and control store.")
        return None, None

    try:
        corr, _ = pearsonr(merged[f'{metric}_trial'], merged[f'{metric}_control'])
        
        obs_distance = np.linalg.norm(
            merged[f'{metric}_trial'].values - merged[f'{metric}_control'].values
        )
        return corr, obs_distance
    
    except Exception as e:
        print(f"Error comparing stores: {e}")
        return None, None
```
